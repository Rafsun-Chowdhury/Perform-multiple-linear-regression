# 📈 Marketing Promotion Sales Prediction – Multiple Linear Regression
This project analyzes the impact of marketing channels (TV, Radio, Social Media, Influencers) on sales using a multiple linear regression model. The focus is on interpreting coefficients and assessing model assumptions.

## Step 1: Import Libraries

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from statsmodels.formula.api import ols
```

## Step 2: Load and Inspect Data

```
data = pd.read_csv('marketing_sales_data.csv')
data.head()
```

## Step 3: Initial Exploration

```
sns.pairplot(data)
```

```
print(data.groupby('TV')['Sales'].mean())
print('')
print(data.groupby('Influencer')['Sales'].mean())
```

```
data = data.dropna(axis=0)
data = data.rename(columns={'Social Media': 'Social_Media'})
```

## Step 4: Build the OLS Model

```
ols_formula = 'Sales ~ C(TV) + Radio'
OLS = ols(formula=ols_formula, data=data)
model = OLS.fit()
print(model.summary())
```

## Step 5: Check Model Assumptions – Linearity

```
fig, axes = plt.subplots(1, 2, figsize=(8,4))
sns.scatterplot(x=data['Radio'], y=data['Sales'], ax=axes[0])
axes[0].set_title('Radio and Sales')
sns.scatterplot(x=data['Social_Media'], y=data['Sales'], ax=axes[1])
axes[1].set_title('Social Media and Sales')
axes[1].set_xlabel('Social Media')
plt.tight_layout()
```

## Step 6: Check Model Assumptions – Normality of Residuals

```
residuals = model.resid
fig, axes = plt.subplots(1, 2, figsize=(8,4))
sns.histplot(residuals, ax=axes[0])
axes[0].set_xlabel('Residual Value')
axes[0].set_title('Histogram of Residuals')
sm.qqplot(residuals, line='s', ax=axes[1])
axes[1].set_title('Normal QQ Plot')
plt.tight_layout()
plt.show()
```

## Step 7: Residuals vs Fitted Values

```
fig = sns.scatterplot(x=model.fittedvalues, y=model.resid)
fig.set_xlabel('Fitted Values')
fig.set_ylabel('Residuals')
fig.set_title('Fitted Values v. Residuals')
fig.axhline(0)
plt.show()
```

## Step 8: Check Multicollinearity (VIF)

```
from statsmodels.stats.outliers_influence import variance_inflation_factor
X = data[['Radio','Social_Media']]
vif = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
df_vif = pd.DataFrame(vif, index=X.columns, columns=['VIF'])
df_vif
```
