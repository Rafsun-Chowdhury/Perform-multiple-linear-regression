# 📊 Campaign Marketing Performance Analysis
This project explores a comprehensive marketing campaign dataset to identify key performance drivers across multiple dimensions such as channel, audience, and campaign type. The analysis includes KPI engineering (CTR, CPC, CPI), exploratory visualizations, and data-driven recommendations for optimizing marketing strategies.

##  Problem Statement
**Goal:** Understand which marketing factors most significantly impact ROI, conversion rate, and engagement. We aim to uncover which channels, audiences, and strategies yield the best performance to support future campaign decisions.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from datetime import datetime
sns.set(style='whitegrid')
```

```
# Load dataset
df = pd.read_csv('marketing_campaign_dataset.csv')

# Clean and transform columns
df['Acquisition_Cost'] = df['Acquisition_Cost'].replace('[\$,]', '', regex=True).astype(float)
df['Duration_Days'] = df['Duration'].str.extract('(\d+)').astype(int)
df['Date'] = pd.to_datetime(df['Date'])

# Feature Engineering
df['CTR'] = df['Clicks'] / df['Impressions']
df['CPC'] = df['Acquisition_Cost'] / df['Clicks']
df['CPI'] = df['Acquisition_Cost'] / df['Impressions']

# Preview
df.head()
```

## 📈 KPI Summary by Channel

```
df.groupby('Channel_Used')[['ROI', 'Conversion_Rate', 'CTR', 'CPC', 'CPI']]
  .mean().sort_values(by='ROI', ascending=False).round(3)
```

```
# Helper function to annotate bar values
def add_value_labels(ax):
    for p in ax.patches:
        height = p.get_height()
        ax.annotate(f'{height:.2f}', 
                    (p.get_x() + p.get_width() / 2., height),
                    ha='center', va='bottom', fontsize=9, color='black')
```

###  Average ROI by Channel

```
plt.figure(figsize=(10,6))
roi_by_channel = df.groupby('Channel_Used')['ROI'].mean().sort_values(ascending=False)
ax = sns.barplot(x=roi_by_channel.index, y=roi_by_channel.values, palette='viridis')
add_value_labels(ax)
plt.title('Average ROI by Marketing Channel')
plt.ylabel('Average ROI')
plt.xlabel('Channel')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

###  Average Conversion Rate by Campaign Type

```
plt.figure(figsize=(10,6))
conversion_by_type = df.groupby('Campaign_Type')['Conversion_Rate'].mean().sort_values(ascending=False)
ax = sns.barplot(x=conversion_by_type.index, y=conversion_by_type.values, palette='Blues_r')
add_value_labels(ax)
plt.title('Average Conversion Rate by Campaign Type')
plt.ylabel('Conversion Rate')
plt.xlabel('Campaign Type')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### 💬 Engagement Score by Audience Group

```
plt.figure(figsize=(10,6))
engagement_by_audience = df.groupby('Target_Audience')['Engagement_Score'].mean().sort_values(ascending=False)
ax = sns.barplot(x=engagement_by_audience.index, y=engagement_by_audience.values, palette='magma')
add_value_labels(ax)
plt.title('Engagement Score by Audience Group')
plt.ylabel('Engagement Score')
plt.xlabel('Target Audience')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

##  Summary & Recommendations

- All marketing channels perform closely in terms of ROI, suggesting a well-balanced campaign distribution.
- Campaign types like Influencer and Email yield higher conversion rates.
- Audience groups show variance in engagement; targeting high-engagement groups can enhance results.
- Monitoring CTR and CPC metrics allows better budget efficiency.

**Recommendations:**
- Prioritize campaign types and channels with strong ROI and conversion.
- Use segmentation insights to customize messaging for engaged audience groups.
- Run A/B tests on underperforming channels to improve results.
